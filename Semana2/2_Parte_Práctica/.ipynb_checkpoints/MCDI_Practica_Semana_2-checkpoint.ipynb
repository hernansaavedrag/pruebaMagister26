# **I. `Scope` y los `Namespaces`**

Antes de abordar el concepto de clases, resulta fundamental comprender los principios de scope y namespace en Python, ya que ambos determinan cómo se organizan, resuelven y acceden los nombres dentro de un programa.

Un namespace puede entenderse como una estructura de mapeo entre identificadores y objetos. En otras palabras, constituye el contexto en el que un nombre queda asociado a una referencia específica dentro del entorno de ejecución. Los namespaces se generan en distintos momentos del ciclo de vida del programa y pueden coexistir con diferentes alcances y duraciones, según la naturaleza del bloque de código en el que se originan.

Durante la ejecución de un programa en Python 3 se crean múltiples namespaces, cada uno vinculado a determinados niveles de scope, lo que define desde qué regiones del código dichos nombres pueden ser accedidos o utilizados. Comprender esta relación es esencial para interpretar correctamente la visibilidad de las variables, evitar ambigüedades en la resolución de nombres y diseñar estructuras de programación más robustas y mantenibles.

# **1. ¿Que es un `Scope`?**

El scope (ámbito) define desde dónde una variable puede ser utilizada.

````python
def saludar():
    mensaje = "Hola"
    print(mensaje)

saludar()

# La variable mensaje NO existe fuera de la función: su scope es local.
# Se captura el error para dejar la demostración sin interrumpir el cuaderno.
try:
    print(mensaje)
except NameError as error:
    print("NameError:", error)
````

La variable mensaje existe solo dentro de la función porque su scope es local.

# **2. ¿Que es un Namespace?**

Un namespace es un espacio donde Python almacena nombres y sus referencias a objetos.
El namespace responde a la pregunta: ¿dónde está almacenado este nombre?

# **3. Relación con Programación Orientada a Objetos**

En Python, las clases y los objetos tienen su propio namespace.


````python
class Persona:
    especie = "Humano"
````

La clase almacena:

Persona

 └── especie

````python
p = Persona()
p.nombre = "Ana"
````

¿Cómo busca Python los atributos?

````python
#Cuando escribimos:
p.especie
````

Python busca:

1. En el namespace del objeto p.

2. Si no lo encuentra, en el namespace de la clase Persona.

3. Luego en las clases padre (si existen).

````python
class Persona:

    def __init__(self, nombre):
        self.nombre = nombre
````

self.nombre guarda el atributo dentro del namespace del objeto.

Por eso

````python
p = Persona("Ana")
print(p.nombre)
````

El namespace es donde se guardan los nombres; el scope determina desde dónde pueden utilizarse. En POO, cada clase y cada objeto tienen su propio namespace, y Python busca atributos y métodos recorriendo esos namespaces.

Continuemos mas a fondo con los conceptos de **Programación Orientada a Objetos**.

# **II. Programación Orientación a Objetos**

La orientación a objetos es un paradigma de programación que permite modelar y organizar soluciones computacionales mediante entidades denominadas objetos, las cuales integran datos y comportamientos relacionados dentro de una misma unidad lógica. Este enfoque favorece una estructuración más modular, reutilizable y comprensible del software, al permitir agrupar propiedades y funcionalidades afines en abstracciones coherentes.

Python es un lenguaje orientado a objetos, y en él todas las estructuras de datos y componentes manipulados durante la ejecución son tratados como objetos. Esto implica que tanto los tipos básicos como las estructuras más complejas poseen identidad, estado y comportamiento asociados.

En este paradigma, la construcción fundamental para definir nuevas abstracciones es la clase, la cual actúa como un modelo o plantilla que describe los atributos y métodos que tendrán los objetos creados a partir de ella. A su vez, los objetos constituyen instancias concretas de una clase y permiten representar entidades, procesos o conceptos del mundo real dentro de un sistema computacional. De este modo, la programación orientada a objetos facilita el diseño de programas más expresivos, escalables y alineados con problemas de mayor complejidad, como los abordados en ciencia de datos e inteligencia artificial.

### 1. Clases

Las clases son el elemento central de la programación orientada a objetos, ya que permiten organizar en una misma estructura tanto los datos como las funciones asociadas a ellos. Cuando se define una clase en Python, se genera un nuevo tipo de dato, a partir del cual pueden crearse distintas instancias, conocidas como objetos. Cada objeto puede almacenar información propia mediante atributos, que representan su estado, y también puede disponer de métodos, es decir, funciones definidas dentro de la clase que permiten manipular o modificar dicho estado.

La sintaxis empleada para definir clases en Python 3 es la siguiente:


```
class <nombre_clase>:
    <sentencia(s)>
```

````python
class Moto:
    pass
````

### 2. Metodos

Dentro de una clase pueden definirse funciones, y cuando una función pertenece a una clase se denomina método. Los métodos representan los comportamientos o acciones de los objetos. En general, todo lo visto sobre funciones también se aplica a los métodos; la principal diferencia es que estos se invocan a través de un objeto o de la clase a la que pertenecen.

````python
class Moto:
    def velocidad_maxima():
        """Este método devuelve la velocidad máxima de la moto."""
        print("Velocidad máxima: ???")
````

````python
moto1 = Moto()
````

### 3. Parámetro `self`

Al definir métodos dentro de una clase, es necesario incluir el parámetro self como primer argumento. Este parámetro representa la instancia concreta del objeto que está utilizando el método en ese momento. Es decir, self permite que el método sepa sobre qué objeto está actuando.

Gracias a self, un método puede acceder a los atributos propios de cada objeto y también invocar otros métodos definidos en la misma clase. Esto es fundamental porque, aunque varios objetos provengan de una misma clase, cada uno puede almacenar valores distintos en sus atributos. Por ello, self permite distinguir la información particular de cada instancia.

Por ejemplo, si una clase Persona tiene un atributo nombre, al escribir self.nombre se está haciendo referencia al nombre del objeto específico que ejecuta el método, no al de la clase en general ni al de otro objeto. De esta manera, cada instancia puede trabajar con sus propios datos de forma independiente.

En síntesis, self es necesario porque conecta los métodos con los datos del objeto que los ejecuta, permitiendo que cada instancia mantenga y manipule su propia información dentro de la clase.

````python
class Moto:
    def velocidad_maxima(self):
        """Este método devuelve la velocidad máxima del coche."""
        print("Velocidad máxima: ???")
````

### 5. Atributos

Dentro de una clase también pueden definirse variables, las cuales, en el contexto de la programación orientada a objetos, reciben el nombre de atributos. Estos atributos permiten almacenar información relacionada con los objetos de la clase.

En Python, los atributos se clasifican en dos tipos principales: atributos de clase y atributos de instancia. Los primeros pertenecen a la clase en general y son compartidos por todos los objetos creados a partir de ella, mientras que los segundos corresponden a cada objeto en particular, por lo que cada instancia puede tener sus propios valores.

Un atributo de clase es una variable definida dentro de la clase que pertenece a la propia clase y, por lo tanto, es compartida por todos los objetos creados a partir de ella. Esto significa que su valor no es exclusivo de una instancia en particular, sino común para todas. Para acceder al valor de un atributo de clase, se utiliza la siguiente sintaxis:

```
<nombre_objeto>.<nombre_atributo_clase>
```

````python
class Moto:
    atributo_clase = 150

    def velocidad_maxima(self):
        """Este método devuelve la velocidad máxima del coche."""
        print("Velocidad máxima:", self.atributo_clase)
````

````python
moto1 = Moto()
````

````python
moto1.velocidad_maxima()
````

````python
moto1.atributo_clase
````

````python
moto2 = Moto()
````

````python
moto2.velocidad_maxima()
````

Un atributo de instancia es una variable que pertenece a un objeto específico, por lo que su valor puede ser distinto en cada instancia de la clase. Estos atributos se definen generalmente dentro del método especial __init__(), conocido como constructor, que se ejecuta al crear un objeto. Para acceder al valor de un atributo de instancia, se utiliza la siguiente sintaxis:

```
<nombre_objeto>.<nombre_atributo_instancia>
```

### 6. Método `__init__()`

El método `__init__()` es un método especial que Python ejecuta automáticamente cada vez que se crea una nueva instancia de una clase. Su nombre tiene dos guiones bajos al inicio y dos al final, ya que forma parte de los métodos especiales del lenguaje, los cuales siguen una convención específica para distinguirlos de los métodos definidos por el usuario.

Este método se conoce como constructor porque permite inicializar el objeto recién creado, asignándole valores a sus atributos. De esta manera, cada instancia puede quedar configurada con información propia desde el momento en que es creada.

````python
class Moto:

    def __init__(self, vel_max, consumo_medio):
        self.vel_max = vel_max
        self.con_medio = consumo_medio

    def velocidad_maxima(self):
        """Este método devuelve la velocidad máxima del coche."""
        print("Velocidad máxima:", self.vel_max)

    def consumo_medio(self):
        print("El consumo medio es:", self.con_medio)
````

````python
moto1 = Moto(200, 7)
````

````python
moto1.velocidad_maxima()
````

````python
moto1.consumo_medio()
````

````python
moto2 = Moto(250, 10)
````

````python
moto2.velocidad_maxima()
````

````python
moto2.consumo_medio()
````

### 7. Herencia

Es un mecanismo de la programación orientada a objetos que permite crear nuevas clases a partir de clases ya existentes, estableciendo así una relación jerárquica entre ellas. Gracias a este proceso, una clase derivada o hija puede reutilizar los atributos y métodos definidos en una clase base o padre. Esto favorece la organización del código y evita repetir funcionalidades comunes.

Además, la clase base define una estructura general o comportamiento común que las clases derivadas comparten. A partir de esa base, cada clase hija puede especializar o adaptar dicho comportamiento, ya sea añadiendo nuevas características o redefiniendo métodos para ofrecer una implementación más específica según sus necesidades.

Este código no muestra herencia directamente, porque solo define una clase llamada Moto y no una clase hija que herede de ella. Sin embargo, sí puede entenderse como una clase base, ya que reúne atributos y métodos generales que podrían ser reutilizados por clases derivadas. La clase define información común de una moto, como modelo, potencia, consumo y kilometraje, además de métodos para mostrar sus especificaciones y calcular el consumo total. También utiliza una propiedad para controlar la actualización de los kilómetros, evitando asignar un valor menor al actual.

````python
class Moto1():
    """Esta clase representa un coche."""

    def __init__(self, modelo, potencia, consumo):
        """Inicializa los atributos de instancia.

        Argumentos posicionales:
        modelo -- string que representa el modelo del coche
        potencia -- int que representa la potencia en cv
        consumo -- int que representa el consumo en litros/100km
        """
        self.modelo = modelo
        self.potencia = potencia
        self.consumo = consumo
        self._km_actuales = 0

    def especificaciones(self):
        """Muestra las especificaciones del coche."""
        print("Modelo:", self.modelo,
              "\nPotencia: {} cv".format(self.potencia),
              "\nConsumo: {} l/100km".format(self.consumo),
              "\nKilometros actuales:", self._km_actuales)

    @property
    def kilometros(self):
        return self._km_actuales

    @kilometros.setter
    def kilometros(self, kilometros):
        """Actualiza los kilometros actuales del coche."""
        if kilometros > self._km_actuales:
            self._km_actuales = kilometros
        else:
            print("ERROR: No se puede establecer un valor de km inferior al actual")

    def consumo_total(self):
        """Muestra el consumo total del coche desde el kilometro 0."""
        consumo_total = (self._km_actuales / 100) * self.consumo
        print("El consumo total es de {} litros".format(consumo_total))
````

````python
mercedes = Moto1("mercedes c200", 180, 7)
````

````python
mercedes.especificaciones()
````

Desde la perspectiva de la herencia, esta clase podría servir como punto de partida para crear clases más específicas, como MotoDeportiva o MotoElectrica, que hereden esta estructura y añadan comportamientos propios.

**¿Qué sucede si quiero representar una moto electrica?**

````python
class Moto1():
    """Esta clase representa un coche."""

    def __init__(self, modelo, potencia, consumo):
        """Inicializa los atributos de instancia.

        Argumentos posicionales:
        modelo -- string que representa el modelo del coche
        potencia -- int que representa la potencia en cv
        consumo -- int que representa el consumo
        """
        self.modelo = modelo
        self.potencia = potencia
        self.consumo = consumo
        self._km_actuales = 0
        self._combustible = "l/100km"

    def especificaciones(self):
        """Muestra las especificaciones del coche."""
        print("Modelo:", self.modelo,
              "\nPotencia: {} cv".format(self.potencia),
              "\nConsumo: {} {}".format(self.consumo, self._combustible),
              "\nKilometros actuales:", self._km_actuales)

    @property
    def kilometros(self):
        return self._km_actuales

    @kilometros.setter
    def kilometros(self, kilometros):
        """Actualiza los kilometros actuales del coche."""
        if kilometros > self._km_actuales:
            self._km_actuales = kilometros
        else:
            print("ERROR: No se puede establecer un valor de km inferior al actual")

    def consumo_total(self):
        """Muestra el consumo total del coche desde el kilometro 0."""
        consumo_total = (self._km_actuales / 100) * self.consumo
        print("El consumo total es de {} litros".format(consumo_total))
````

````python
moto_electrica = Moto1("E1 model 3", 300, 15) # consumo: 15Kwh/100km
````

````python
moto_electrica.especificaciones()
````

En este punto podría definirse un método setter para modificar el valor del combustible; sin embargo, incorporar este tipo de lógica dentro de la clase base podría hacerla demasiado extensa, compleja y difícil de mantener. Cuando una clase concentra numerosos métodos y validaciones para atender casos particulares, aumenta el riesgo de generar código repetido y poco claro. Por esta razón, en situaciones como esta resulta más adecuado recurrir a la herencia, dejando en la clase base únicamente los elementos comunes y trasladando los comportamientos específicos a clases derivadas. De este modo, se consigue un diseño más ordenado, reutilizable y fácil de comprender.

````python
class Moto_electrica(Moto1):
    """Esta clase representa un coche electrico."""

    def __init__(self, modelo, potencia, consumo):
        """Inicializa los atributos de la clase padre."""
        super().__init__(modelo, potencia, consumo)
        self._combustible = "KWh/100km"
````

````python
moto_E1 = Moto_electrica("E1 model 3", 300, 15)
````

````python
moto_E1.especificaciones()
````

### 8. Definición de atributos y métodos propios en la clase hija.


Además de heredar los elementos de la clase padre, una clase hija puede ampliar su comportamiento incorporando nuevos atributos y métodos propios. Esto permite especializar la clase derivada, añadiendo características o funcionalidades particulares que no estaban definidas en la clase base.

````python
class moto_electrica1(Moto1):
    """Esta clase representa un coche electrico."""

    def __init__(self, modelo, potencia, consumo, capacidad_bateria):
        """Inicializa los atributos de la clase padre."""
        super().__init__(modelo, potencia, consumo)
        self._combustible = "KWh/100km"
        self._capacidad_bateria = capacidad_bateria

    def detalles_bateria(self):
        """Muestra los detalles de la bateria del coche electrico."""
        print("El tamaño de la batería es: {} KWh".format(self._capacidad_bateria))
````

````python
E11 = moto_electrica1("E11 model 3", 300, 15, 50)
````

````python
E11.especificaciones()
````

````python
E11.detalles_bateria()
````

### 9. Sobrescritura de métodos de la clase padre

En algunos casos, un método heredado de la clase padre puede no ajustarse completamente al comportamiento que se desea en la clase hija. Cuando esto ocurre, es posible sobrescribirlo en la clase derivada, definiendo una nueva implementación con el mismo nombre. De esta manera, la clase hija puede adaptar ese comportamiento a sus propias necesidades.

````python
class moto_electrica2(Moto1):
    """Esta clase representa un coche electrico."""

    def __init__(self, modelo, potencia, consumo, capacidad_bateria):
        """Inicializa los atributos de la clase padre."""
        super().__init__(modelo, potencia, consumo)
        self._combustible = "KWh/100km"
        self._capacidad_bateria = capacidad_bateria

    def detalles_bateria(self):
        """Muestra los detalles de la bateria del coche electrico."""
        print("El tamaño de la batería es: {} KWh".format(self._capacidad_bateria))

    def consumo_total(self):
        """Muestra el consumo total del coche desde el kilometro 0."""
        consumo_total = (self._km_actuales / 100) * self.consumo
        print("El consumo total es de {} kWh".format(consumo_total))
````

````python
E2 = moto_electrica2("E2 model 3", 300, 15, 50)
````

### 10. Objetos dentro de una clase

En algunos casos, ciertas propiedades de una clase pueden tener la suficiente importancia o complejidad como para representarse mediante una clase propia. Cuando esto ocurre, en lugar de usar un valor simple, es posible asignar un objeto de esa segunda clase como atributo de la primera. Esto permite modelar la información de una forma más organizada, clara y cercana a la realidad.

````python
class Bateria:
    """Esta clase va a representar una bateria de un coche electrico."""

    def __init__(self, capacidad, tipo_pila, num_pilas, peso):
        self._capacidad = capacidad
        self._tipo_pila = tipo_pila
        self._num_pilas = num_pilas
        self._peso = peso

    def especificaciones(self):
        print("Capacidad {} kWh".format(self._capacidad),
             "\nTipo de pilas:", self._tipo_pila,
             "\nNumero de pilas:", self._num_pilas,
             "\nPeso de la bateria: {} kg".format(self._peso))
````

````python
bateria_tesla_modelS = Bateria(80, 2170, 203_136, 480)
````

````python
bateria_tesla_modelS.especificaciones()
````

````python
class CocheElectrico(Moto1):
    """Esta clase representa un coche electrico."""

    def __init__(self, modelo, potencia, consumo, bateria):
        """Inicializa los atributos de la clase padre."""
        super().__init__(modelo, potencia, consumo)
        self._combustible = "KWh/100km"
        self._bateria = bateria

    def detalles_bateria(self):
        """Muestra los detalles de la bateria del coche electrico."""
        self._bateria.especificaciones()

    def consumo_total(self):
        """Muestra el consumo total del coche desde el kilometro 0."""
        consumo_total = (self._km_actuales / 100) * self.consumo
        print("El consumo total es de {} kWh".format(consumo_total))
````

````python
tesla = CocheElectrico("tesla model S", 450, 20, bateria_tesla_modelS)
````

# **IV. Eficiencia: medir tiempo y memoria**

Hasta acá el criterio fue que el código funcione. Ahora se agrega otro: **cuánto cuesta
ejecutarlo**.

En la Fase 3 no basta con afirmar que una solución es más rápida. Hay que medirlo, mostrar
la cifra y explicar qué significa. Esta sección enseña cómo hacerlo.

### 1. Las tres herramientas

| Herramienta | Qué mide | Cuándo usarla |
|---|---|---|
| `time.perf_counter()` | Tiempo transcurrido de un intervalo | Etapas de un proceso |
| `timeit.timeit()` | Tiempo acumulado repitiendo la operación | Fragmentos breves |
| `tracemalloc` | Memoria que asigna Python | Cuando el volumen es grande |

**No usen `time.time()`.** Tiene menos resolución y se ve afectado por ajustes del reloj
del sistema. Está pensado para saber qué hora es, no para medir intervalos cortos.

````python
import time
import timeit
import tracemalloc
import random

# Se fija la semilla para que los resultados sean reproducibles
random.seed(42)

print("Herramientas de medición cargadas.")
````

### 2. `perf_counter`: medir un intervalo

Se toma el tiempo antes, se ejecuta la operación, se toma el tiempo después y se resta.

````python
inicio = time.perf_counter()

# La operación que se quiere medir
total = 0
for i in range(1_000_000):
    total = total + i

transcurrido = time.perf_counter() - inicio

print("Resultado:", total)
print("Tiempo   :", round(transcurrido, 5), "segundos")
````

### 3. Un caso real con nuestras motos

Supongan un concesionario con muchas motos y la necesidad de buscar una por su modelo.
Hay dos formas de resolverlo, y tienen costos muy distintos.

````python
class MotoInventario:
    """Moto simplificada, solo para el inventario del concesionario."""

    def __init__(self, modelo, potencia, precio):
        self.modelo = modelo
        self.potencia = potencia
        self.precio = precio

    def __repr__(self):
        return f"MotoInventario({self.modelo}, {self.potencia} cv)"


def crear_inventario(cantidad):
    """Genera una lista de motos con modelos únicos."""
    return [MotoInventario(f"modelo-{i:05d}",
                           random.randint(100, 300),
                           random.randint(2_000_000, 12_000_000))
            for i in range(cantidad)]


inventario = crear_inventario(50_000)
print("Motos en el inventario:", len(inventario))
print("Primera:", inventario[0])
````

#### Forma A: recorrer la lista hasta encontrar

Se revisa moto por moto. Si la buscada está al final, hay que mirarlas todas.

````python
def buscar_recorriendo(inventario, modelo):
    """Recorre la lista comparando uno por uno."""
    for moto in inventario:
        if moto.modelo == modelo:
            return moto
    return None


encontrada = buscar_recorriendo(inventario, "modelo-49999")
print("Encontrada:", encontrada)
````

#### Forma B: construir un índice y buscar en él

Se arma una sola vez un diccionario que asocia cada modelo con su moto. Después, cada
búsqueda es prácticamente instantánea, sin importar cuántas motos haya.

````python
def construir_indice(inventario):
    """Arma un diccionario modelo -> moto. Cuesta una pasada."""
    return {moto.modelo: moto for moto in inventario}


indice = construir_indice(inventario)


def buscar_en_indice(indice, modelo):
    """Busca directamente por la llave."""
    return indice.get(modelo)


encontrada = buscar_en_indice(indice, "modelo-49999")
print("Encontrada:", encontrada)
````

### 4. Comparar las dos formas

Antes de comparar tiempos hay que comprobar que **ambas devuelven lo mismo**. Una versión
más rápida que entrega otra respuesta no es una optimización: es un error.

````python
# Comprobación de equivalencia
modelo_buscado = "modelo-49999"
assert (buscar_recorriendo(inventario, modelo_buscado) is
        buscar_en_indice(indice, modelo_buscado)), "Las dos formas no coinciden"
print("Las dos formas devuelven exactamente la misma moto.\n")

# Medición con timeit: repite la operación y entrega el tiempo acumulado
repeticiones = 100
t_lista = timeit.timeit(lambda: buscar_recorriendo(inventario, modelo_buscado),
                        number=repeticiones)
t_indice = timeit.timeit(lambda: buscar_en_indice(indice, modelo_buscado),
                         number=repeticiones)

print(f"Recorriendo la lista : {t_lista:.5f} s")
print(f"Buscando en el índice: {t_indice:.7f} s")
print(f"\nEl índice es {t_lista / t_indice:,.0f} veces más rápido")
````

**Atención al costo de construir el índice.** La búsqueda en el diccionario es mucho más
rápida, pero armarlo tiene un costo que hay que pagar una vez.

Si van a buscar **una sola vez**, recorrer la lista puede ser preferible. Si van a buscar
**muchas veces**, el índice se paga solo. Esa es exactamente la clase de decisión que la
Fase 3 pide justificar con cifras.

````python
inicio = time.perf_counter()
construir_indice(inventario)
costo_indice = time.perf_counter() - inicio

costo_una_busqueda_lista = t_lista / repeticiones
busquedas_equilibrio = costo_indice / costo_una_busqueda_lista

print(f"Construir el índice cuesta      : {costo_indice:.5f} s")
print(f"Una búsqueda en la lista cuesta : {costo_una_busqueda_lista:.5f} s")
print(f"\nEl índice conviene a partir de {busquedas_equilibrio:.0f} búsquedas.")
````

### 5. Cómo crece el costo con el tamaño

Una medición sobre un solo tamaño no dice mucho. Lo que revela la **complejidad** es ver
cómo cambia el tiempo cuando el volumen crece.

````python
print(f"{'motos':>10} {'lista (ms)':>13} {'índice (ms)':>13} {'razón':>10}")
print("-" * 50)

for cantidad in [1_000, 5_000, 20_000, 50_000]:
    inv = crear_inventario(cantidad)
    idx = construir_indice(inv)
    ultimo = f"modelo-{cantidad - 1:05d}"

    t_l = timeit.timeit(lambda: buscar_recorriendo(inv, ultimo), number=50) / 50
    t_i = timeit.timeit(lambda: buscar_en_indice(idx, ultimo), number=50) / 50

    print(f"{cantidad:>10,} {t_l * 1000:>13.4f} {t_i * 1000:>13.6f} {t_l / t_i:>10,.0f}")
````

**Lean la última columna.** La ventaja del índice **crece** con el tamaño, mientras que el
tiempo de la lista aumenta de forma proporcional al número de motos.

Eso es lo que significa que una solución sea de costo **lineal**, escrito O(n), y la otra
de costo **constante**, escrito O(1). El primero depende del tamaño; el segundo no.

### 6. Medir la memoria con `tracemalloc`

El tiempo no es el único costo. El índice es más rápido porque **gasta memoria**: guarda
una estructura adicional con todas las llaves.

````python
def medir(funcion, *args, **kwargs):
    """Ejecuta la función y devuelve resultado, segundos y memoria pico en MB.

    *args recoge los argumentos posicionales en una tupla y **kwargs los
    argumentos con nombre en un diccionario. El asterisco es lo que hace el
    trabajo; los nombres args y kwargs son solo convención.
    """
    tracemalloc.start()
    inicio = time.perf_counter()
    resultado = funcion(*args, **kwargs)
    transcurrido = time.perf_counter() - inicio
    _, pico = tracemalloc.get_traced_memory()
    tracemalloc.stop()
    return resultado, transcurrido, pico / 1024 / 1024


_, seg_busqueda, mem_busqueda = medir(buscar_recorriendo, inventario, "modelo-49999")
_, seg_indice, mem_indice = medir(construir_indice, inventario)

print(f"{'operación':<26} {'segundos':>10} {'memoria MB':>12}")
print("-" * 50)
print(f"{'buscar recorriendo':<26} {seg_busqueda:>10.5f} {mem_busqueda:>12.3f}")
print(f"{'construir el índice':<26} {seg_indice:>10.5f} {mem_indice:>12.3f}")
````

**Ahí está el intercambio.** El índice gana en tiempo y pierde en memoria. Cuando en el
informe justifiquen una decisión de eficiencia, corresponde mencionar **las dos
dimensiones**: el descriptor de la rúbrica habla de complejidad temporal y espacial.

### 7. Tres reglas para que la medición valga

1. **Repitan y conserven el tiempo menor**, no el promedio. Los valores altos suelen
   deberse a que el computador estaba haciendo otra cosa, no a su código.
2. **Midan sobre el tamaño real** de su conjunto. En cien elementos todo es instantáneo y
   la comparación no distingue nada.
3. **Comprueben que las versiones dan el mismo resultado**, con un `assert` antes de
   comparar tiempos.

**Lo que distingue el nivel superior:** que la medición **cambie una decisión**. Medir y
no cambiar nada es un ejercicio. Medir, ver dónde está el costo y reescribir esa parte es
optimización.

# **V. Patrones de diseño**

Un **patrón de diseño** es una solución conocida a un problema que se repite. No es código
para copiar: es una forma de organizar las relaciones entre clases.

Lo importante: **varios de estos patrones ya los usaron sin saber su nombre.**

| Patrón | El problema que resuelve |
|---|---|
| **Strategy** | Varias formas de hacer lo mismo, que hay que poder intercambiar |
| **Factory** | Elegir qué objeto construir según una condición |
| **Observer** | Registrar lo que ocurre sin que quien ejecuta sepa quién anota |
| **Singleton** | Que exista una sola configuración compartida por todo el programa |

## 1. Strategy: varias formas de hacer lo mismo

**El problema.** El consumo de una moto no se calcula igual en ciudad, en carretera o en
uso mixto. Si cada cálculo es un método distinto dentro de la clase, agregar una forma
nueva obliga a modificar la clase.

**La solución.** Separar el **qué** se hace, calcular el consumo, del **cómo** se hace.
El cálculo se entrega como un objeto aparte que la moto recibe.

````python
class EstrategiaConsumo:
    """Contrato común de todas las formas de calcular el consumo."""

    etiqueta = "sin definir"

    def calcular(self, consumo_base, kilometros):
        raise NotImplementedError("Cada estrategia debe implementarlo.")


class ConsumoUrbano(EstrategiaConsumo):
    """En ciudad el consumo sube por las detenciones."""

    etiqueta = "urbano"

    def calcular(self, consumo_base, kilometros):
        return (consumo_base * 1.30) * kilometros / 100


class ConsumoCarretera(EstrategiaConsumo):
    """En carretera el consumo baja por la velocidad constante."""

    etiqueta = "carretera"

    def calcular(self, consumo_base, kilometros):
        return (consumo_base * 0.85) * kilometros / 100


class ConsumoMixto(EstrategiaConsumo):
    """Promedio ponderado entre ciudad y carretera."""

    etiqueta = "mixto"

    def __init__(self, proporcion_urbana=0.5):
        self.proporcion_urbana = proporcion_urbana

    def calcular(self, consumo_base, kilometros):
        urbano = ConsumoUrbano().calcular(consumo_base, kilometros)
        carretera = ConsumoCarretera().calcular(consumo_base, kilometros)
        p = self.proporcion_urbana
        return urbano * p + carretera * (1 - p)
````

````python
class MotoConEstrategia:
    """La moto NO sabe calcular el consumo: sabe cuándo pedirlo.

    El cómo lo aporta la estrategia que recibe en el constructor.
    """

    def __init__(self, modelo, consumo_base, estrategia=None):
        self.modelo = modelo
        self.consumo_base = consumo_base
        self.estrategia = estrategia or ConsumoMixto()

    def consumo_en(self, kilometros):
        return self.estrategia.calcular(self.consumo_base, kilometros)


# Cambiar de estrategia no modifica ni una línea de la clase MotoConEstrategia
moto = MotoConEstrategia("Honda CB500", consumo_base=4.2)

print(f"{'estrategia':<14} {'litros en 300 km':>18}")
print("-" * 34)
for estrategia in [ConsumoUrbano(), ConsumoCarretera(), ConsumoMixto(0.6)]:
    moto.estrategia = estrategia
    print(f"{estrategia.etiqueta:<14} {moto.consumo_en(300):>18.2f}")
````

**Fíjense en lo que acaba de pasar.** Comparar tres formas de calcular costó un bucle de
tres líneas, y la clase `MotoConEstrategia` no cambió.

**En su proyecto de datos** este es el patrón que corresponde a comparar estrategias de
imputación: mediana, media o por grupo. Es exactamente la misma estructura.

## 2. Factory: decidir qué objeto construir

**El problema.** El tipo de moto que hay que crear depende de una condición. Esa decisión
suele quedar como una cadena de `if` repetida en varios lugares del programa.

**La solución.** Concentrar la decisión en una sola función que devuelve el objeto ya
construido. El resto del código pide una moto y no necesita saber cuál recibió.

````python
class MotoBencina:
    tipo = "bencina"

    def __init__(self, modelo, potencia):
        self.modelo = modelo
        self.potencia = potencia

    def describir(self):
        return f"{self.modelo}: {self.potencia} cv, motor a bencina"


class MotoElectricaSimple:
    tipo = "eléctrica"

    def __init__(self, modelo, potencia, autonomia=120):
        self.modelo = modelo
        self.potencia = potencia
        self.autonomia = autonomia

    def describir(self):
        return f"{self.modelo}: {self.potencia} cv, {self.autonomia} km de autonomía"


class MotoHibrida:
    tipo = "híbrida"

    def __init__(self, modelo, potencia):
        self.modelo = modelo
        self.potencia = potencia

    def describir(self):
        return f"{self.modelo}: {self.potencia} cv, motor híbrido"
````

````python
def crear_moto(tipo, modelo, potencia, **extras):
    """Fábrica: devuelve la clase que corresponde al tipo indicado.

    Lanza ValueError ante un tipo desconocido en lugar de devolver None:
    un fallo silencioso acá dejaría el programa sin moto y sin aviso.
    """
    tipo = tipo.strip().lower()

    if tipo == "bencina":
        return MotoBencina(modelo, potencia)
    if tipo in ("electrica", "eléctrica"):
        return MotoElectricaSimple(modelo, potencia,
                                   extras.get("autonomia", 120))
    if tipo == "hibrida":
        return MotoHibrida(modelo, potencia)

    raise ValueError(
        f"Tipo de moto desconocido: '{tipo}'. "
        "Tipos válidos: bencina, electrica, hibrida."
    )


# Un catálogo en forma de datos gobierna qué objetos se crean
catalogo = [
    {"tipo": "bencina", "modelo": "Honda CB500", "potencia": 47},
    {"tipo": "electrica", "modelo": "Zero SR/F", "potencia": 110, "autonomia": 260},
    {"tipo": "hibrida", "modelo": "Piaggio MP3", "potencia": 26},
]

for ficha in catalogo:
    datos = dict(ficha)
    moto = crear_moto(datos.pop("tipo"), datos.pop("modelo"),
                      datos.pop("potencia"), **datos)
    print(f"[{moto.tipo:<10}] {moto.describir()}")

# Un tipo no contemplado falla con un mensaje claro
try:
    crear_moto("nuclear", "X1", 500)
except ValueError as error:
    print("\nValueError:", error)
````

**En su proyecto de datos** este patrón corresponde a decidir qué transformación aplicar
según el rol de cada variable. Con Factory, el diccionario de variables que armaron en la
Fase 1 deja de ser un documento y pasa a gobernar el código.

## 3. Observer: registrar sin estorbar

**El problema.** Queremos llevar un registro de lo que le pasa a una moto, pero sin que la
clase Moto tenga que saber quién lleva ese registro ni cuántos registros hay.

**La solución.** La moto **avisa** cuando ocurre algo, y quien quiera registrar se
suscribe. La moto no conoce a sus observadores.

````python
class Bitacora:
    """Observador que acumula lo ocurrido y lo entrega como lista."""

    def __init__(self):
        self.eventos = []

    def notificar(self, moto, evento, detalle):
        self.eventos.append({"moto": moto, "evento": evento, "detalle": detalle})

    def resumen(self):
        for i, e in enumerate(self.eventos, start=1):
            print(f"  {i}. [{e['moto']}] {e['evento']}: {e['detalle']}")


class AlertaMantencion:
    """Otro observador, con una regla: avisa al superar un umbral."""

    def __init__(self, umbral_km=10_000):
        self.umbral_km = umbral_km

    def notificar(self, moto, evento, detalle):
        if evento == "kilometraje" and detalle >= self.umbral_km:
            print(f"  [ALERTA] {moto} superó los {self.umbral_km:,} km: toca mantención")
````

````python
class MotoObservable:
    """Moto que avisa a sus observadores cada vez que ocurre algo."""

    def __init__(self, modelo):
        self.modelo = modelo
        self._kilometros = 0
        self._observadores = []       # interno: no se manipula desde fuera

    def suscribir(self, observador):
        self._observadores.append(observador)
        return self                   # permite encadenar suscripciones

    def _avisar(self, evento, detalle):
        """Recorre los observadores sin saber quiénes son."""
        for observador in self._observadores:
            observador.notificar(self.modelo, evento, detalle)

    def recorrer(self, kilometros):
        self._kilometros = self._kilometros + kilometros
        self._avisar("recorrido", f"{kilometros} km")
        self._avisar("kilometraje", self._kilometros)


bitacora = Bitacora()
moto = MotoObservable("Honda CB500")
moto.suscribir(bitacora).suscribir(AlertaMantencion(umbral_km=10_000))

print("Recorridos:")
moto.recorrer(4_500)
moto.recorrer(6_200)

print("\nBitácora producida por la propia ejecución:")
bitacora.resumen()
````

**Lo relevante es lo que la clase `MotoObservable` no sabe.** No conoce a la `Bitacora` ni
a la `AlertaMantencion`. Solo avisa. Se puede agregar un observador nuevo sin tocar ni una
línea de la moto.

**En su proyecto de datos** este patrón convierte la bitácora de decisiones en producto de
la ejecución. Deja de escribirse a mano después de correr el pipeline, así que no puede
quedar desactualizada.

## 4. Singleton: una sola configuración

**El problema.** Parámetros como la semilla o las unidades de medida se necesitan en varias
partes del programa. Si cada parte construye su propia configuración, dos zonas del código
pueden trabajar con valores distintos sin que nadie lo note.

**La advertencia.** Singleton introduce estado global, que dificulta las pruebas. Úsenlo
solo para configuración, y pasen el resto de los parámetros de forma explícita.

````python
class Configuracion:
    """Una sola instancia compartida por todo el programa."""

    _instancia = None

    def __new__(cls, *args, **kwargs):
        # __new__ se ejecuta ANTES que __init__ y decide si se crea un objeto
        if cls._instancia is None:
            cls._instancia = super().__new__(cls)
            cls._instancia._iniciada = False
        return cls._instancia

    def __init__(self, unidad="km", semilla=42):
        # Esta guarda evita que una segunda llamada reinicie el estado.
        # Es el error clásico al implementar Singleton.
        if self._iniciada:
            return
        self.unidad = unidad
        self.semilla = semilla
        self.parametros = {}
        self._iniciada = True

    def definir(self, clave, valor):
        self.parametros[clave] = valor
        return self


a = Configuracion(unidad="km", semilla=42)
b = Configuracion(unidad="millas", semilla=99)   # se ignora: la instancia ya existe

a.definir("umbral_mantencion", 10_000)

print("¿Son el mismo objeto?", a is b)
print("Unidad vista desde b :", b.unidad)
print("Parámetros desde b   :", b.parametros)
````

## 5. Resumen: qué patrón usar y cuándo

| Patrón | El problema que resuelve | Dónde aparece en un proyecto de datos |
|---|---|---|
| **Strategy** | Varias formas de hacer lo mismo | Comparar estrategias de imputación |
| **Factory** | Elegir qué construir según una condición | Decidir la transformación según el rol de la variable |
| **Observer** | Registrar sin que el ejecutor sepa quién anota | La bitácora de decisiones del pipeline |
| **Singleton** | Una sola configuración compartida | La semilla y las rutas del proyecto |

**No usen los cuatro.** Usen el que su proyecto necesita y expliquen por qué. Un patrón
aplicado donde no hace falta agrega complejidad sin beneficio, y eso también se evalúa.

### Cómo se declara un patrón en el informe

Nombrarlo no basta. Escriban tres cosas: **qué problema existía**, **cómo lo resuelve el
patrón** y **qué habría pasado sin él**. Por ejemplo:

> Aplicamos Strategy en el cálculo del consumo. Antes, agregar una forma nueva obligaba a
> modificar la clase. Ahora la estrategia se pasa como parámetro y la clase permanece
> intacta, lo que nos permitió comparar tres alternativas sin reescribir código.
